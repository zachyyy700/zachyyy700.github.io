# Song Frequency Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import IPython.display as ipd
import ipywidgets as widgets
import librosa
import os
import soundfile as sf
```

```python
song_path = os.path.join('..', '..', 'data', 'song1.wav')
# Load the audio file 'audio.wav' with default sample rate
song1, sr = librosa.load(song_path)
```

```python
# chunk duration of 10 seconds
chunk_duration = 10
chunk_samples = int(chunk_duration * sr)

chunks = [song1[i:i + chunk_samples] for i in range(0, len(song1), chunk_samples)]

chunk_path = os.path.join('..', '..', 'data', 'song1_chunks')
```

```python
for i, chunk in enumerate(chunks):
    output_file = os.path.join(chunk_path, f"chunk_{i}.wav")
    sf.write(output_file, chunk, sr)
```

# Picking Chunks

## Chunk 0

```python
chunk0_path = os.path.join('..', '..', 'data', 'song1_chunks', 'chunk_0.wav')
chunk, sr = librosa.load(chunk0_path)
chunk_widget = ipd.Audio(open(chunk0_path, 'rb').read())
```

```python
chunk_widget
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10,6))

librosa.display.waveshow(ax=ax1, y=chunk, sr=sr, color='blue')

# Short-time Fourier transform
D = librosa.stft(chunk)
S = librosa.amplitude_to_db(np.abs(D), ref=np.max)

librosa.display.specshow(ax=ax2, data=S, x_axis='time', y_axis='log')
plt.show()
```
